```python
from datasets import load_dataset
import json
```

```python
DATASET_PATH = 'casperhansen/pmc-oa-markdown' 
save_file = 'pmc_articles.jsonl'
articles_limit = 15000
count = 0
```

```python
try:
    # 2. Pass the STRING constant into the function
    # And save the result to a NEW variable name
    loaded_data = load_dataset(DATASET_PATH, streaming=True)
    
    # 3. Use 'w' as a string
    with open(save_file, 'w', encoding='utf-8') as f:
        for article in loaded_data["train"]:
            text_data = article.get("text")
            record = {"text": text_data}
            
            f.write(json.dumps(record) + "\n")
            count += 1
            print(f"Successfully saved {count} articles")
            
            if articles_limit and count >= articles_limit:
                break
                
except Exception as e:
    print(f"Error occurred: {e}")
```
